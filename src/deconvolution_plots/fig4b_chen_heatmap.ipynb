# Fig 4B – Chen dataset: clustered heatmap of MCT fractions

Min-max scaled MCT fractions per sample, annotated with collection center and phenotype. As in the original notebook, two files are written: the heatmap without legends (`fig4b_chen_heatmap`) and the same heatmap with legends (`fig4b_chen_heatmap_legend`); the published panel places the legends of the second below the first.

```python
# {dataset}: Flomics_2, chalasani_merged, chen, ibarra, moufarrej, reggiardo, roskams, tao, toden, zhu, decru
FRACTIONS_FILE = "/mnt/efs/home/gasole/projects/FL-cfRNAmeta/script/2026_10_26_scripts_review/inputs_test/deconv_results{dataset}/dec_cfRNAmeta_fractions_results.csv"
SAMPLE_LIST_FILE = "/mnt/efs/home/gasole/projects/FL-cfRNAmeta/script/2026_10_26_scripts_review/inputs_test/deconv_results{dataset}/sample_list.txt"

OUTPUT_DIR = "../../output/deconvolution"
```

```python
import sys
from pathlib import Path

import pandas as pd

sys.path[:0] = [".", ".."]
import deconv_common

OUT = Path(OUTPUT_DIR)
deconv_common.set_fonts()
```

```python
metadata = deconv_common.load_metadata()
labels = deconv_common.load_labels()
per_dataset, cell_types = deconv_common.load_deconvolution(metadata, FRACTIONS_FILE, SAMPLE_LIST_FILE)
mcts = deconv_common.select_mcts(per_dataset, cell_types)
print("major cell types:", mcts)
data = deconv_common.add_plot_batches(pd.concat(per_dataset.values()), labels)
```

```python
import heatmap_plot
import matplotlib.pyplot as plt

chen = per_dataset["chen"].copy()
chen["Collection center"] = chen["collection_center_short_name"]
chen["Phenotype"] = chen["phenotype"]
names = {c: c[0].upper() + c[1:] for c in mcts}
chen = chen.rename(columns=names)
colors = {"Phenotype": deconv_common.HEATMAP_COLORS["phenotype"]}

plt.rcParams["legend.fontsize"] = 5
plt.rcParams["legend.title_fontsize"] = 5

common = dict(metadata_colors_mapping=colors, remove_all_zero=True, row_cluster=True,
              rescale_values=True, row_height=2, fig_height_inc=2,
              heatmap_label_fontsize=5, annotation_fontsize=5)
heatmap_plot.create_heatmap(chen, list(names.values()), ["Collection center", "Phenotype"],
                            output=str(OUT / "fig4b_chen_heatmap"), fig_width_inc=4.5,
                            show_legend=False, **common)
plt.close("all")
heatmap_plot.create_heatmap(chen, list(names.values()), ["Collection center", "Phenotype"],
                            output=str(OUT / "fig4b_chen_heatmap_legend"), fig_width_inc=4.1,
                            show_legend=True, colorbar_label="Cell type fraction", **common)
```
